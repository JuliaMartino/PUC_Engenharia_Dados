# Música em Contexto: Music Brainz + Spotify

## Análise

Aqui é feita a análise da tabela **contexto_gold** criada durante esse trabalho, com o objetivo de responder às perguntas do **Objetivo** do estudo.

Importando a tabela **contexto_gold** como **df_contexto_gold**

```python
from pyspark.sql import functions as F

SEED = 42
```

```python
df_contexto_gold = spark.read.table("mvp.gold.contexto_gold")

display(df_contexto_gold.limit(100))
```

## Quais Países Têm as Músicas mais Dançantes?

```python
danceability_pais = (
    df_contexto_gold
    .groupBy("country")
    .agg(
        F.avg("danceability").alias("danceability_media"),
        F.count("*").alias("n_musicas")
    )
    .filter(F.col("n_musicas") >= 100)
    .orderBy(F.desc("danceability_media"))
    .limit(10)
)

display(danceability_pais)
```

Os quatro países que tem músicas mais dançantes são: Jamaica (JM), República Dominicana (DO), Ilhas Virgens Americanas (VI) e Costa do Marfim (CI).

### Quais Décadas Têm as Músicas mais Dançantes?

Primeiro eu crio a **df_contexto_decada** com a coluna derivada **decada**

```python
df_contexto_decada = df_contexto_gold.withColumn(
    "decada",
    (F.floor(F.year("album_release_date") / 10) * 10).cast("int")
)

display(
    df_contexto_decada
    .sample(withReplacement=False, fraction=0.03, seed=SEED)
    .limit(200)
)
```

```python
danceability_decada = (
    df_contexto_decada
    .groupBy("decada")
    .agg(
        F.avg("danceability").alias("danceability_media"),
        F.count("*").alias("n_musicas")
    )
    .orderBy("decada")
)

display(danceability_decada)
```

A média de dançabilidade é muito uniforme entre as décadas

## Quais Países Têm Músicas Mais Instrumentais?

```python
instrumentalness_pais = (
    df_contexto_gold
    .groupBy("country")
    .agg(
        F.avg("instrumentalness").alias("instrumentalness_media"),
        F.count("*").alias("n_musicas")
    )
    .filter(F.col("n_musicas") >= 100)
    .orderBy(F.desc("instrumentalness_media"))
    .limit(10)
)

display(instrumentalness_pais)
```

Os seis países com uma média mais alta de instrumentalness são: Geórgia (GE), Malta (MT), Jersey (JE), Emirados Árabes(AE), Azerbaijão (AZ) e Tunísia (TN).
